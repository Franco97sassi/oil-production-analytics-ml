```python
columnas_ml = [
    "mes",
    "prod_pet",
    "iny_agua",
    "iny_gas",
    "tef",
    "tipoextraccion",
    "tipoestado",
    "tipopozo",
    "provincia",
    "cuenca"
]

df_ml = df[columnas_ml].copy()

df_ml.head()
```

```python
df_ml = df_ml[df_ml["prod_pet"] > 0].copy()

print("Registros para ML:", len(df_ml))
print("\nNulos:")
print(df_ml.isnull().sum())

print("\nProducción de petróleo:")
print(df_ml["prod_pet"].describe())
```

```python
X = df_ml.drop(columns=["prod_pet"])
y = df_ml["prod_pet"]

print("Features:", X.columns.tolist())
print("X:", X.shape)
print("y:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

columnas_numericas = [
    "mes",
    "iny_agua",
    "iny_gas",
    "tef"
]

columnas_categoricas = [
    "tipoextraccion",
    "tipoestado",
    "tipopozo",
    "provincia",
    "cuenca"
]

preprocessor = ColumnTransformer(
    transformers=[
        ("categoricas", OneHotEncoder(handle_unknown="ignore"), columnas_categoricas)
    ],
    remainder="passthrough"
)

print("Columnas numéricas:", columnas_numericas)
print("Columnas categóricas:", columnas_categoricas)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

modelo = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=100,
        max_depth=15,
        random_state=42,
        n_jobs=-1
    ))
])

print("Entrenando modelo...")

modelo.fit(X_train, y_train)

print("Modelo entrenado correctamente.")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Predicciones sobre datos que el modelo nunca vio
y_pred = modelo.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred, alpha=0.3)

limite = max(y_test.max(), y_pred.max())
plt.plot([0, limite], [0, limite], "--")

plt.xlabel("Producción real")
plt.ylabel("Producción predicha")
plt.title("Producción real vs predicha")

plt.tight_layout()
plt.show()
```

```python
y_train_log = np.log1p(y_train)

modelo_log = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=100,
        max_depth=15,
        random_state=42,
        n_jobs=-1
    ))
])

print("Entrenando modelo con target logarítmico...")

modelo_log.fit(X_train, y_train_log)

print("Modelo log entrenado correctamente.")
```

```python
# Predicción del modelo logarítmico
y_pred_log = modelo_log.predict(X_test)

# Volvemos de escala logarítmica a producción real
y_pred_log_real = np.expm1(y_pred_log)

# Métricas
mae_log = mean_absolute_error(y_test, y_pred_log_real)
rmse_log = np.sqrt(mean_squared_error(y_test, y_pred_log_real))
r2_log = r2_score(y_test, y_pred_log_real)

print("=== MODELO ORIGINAL ===")
print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)

print("\n=== MODELO LOG ===")
print("MAE:", mae_log)
print("RMSE:", rmse_log)
print("R²:", r2_log)
```

```python
from sklearn.dummy import DummyRegressor

baseline = DummyRegressor(strategy="mean")
baseline.fit(X_train, y_train)

y_pred_baseline = baseline.predict(X_test)

mae_baseline = mean_absolute_error(y_test, y_pred_baseline)
rmse_baseline = np.sqrt(mean_squared_error(y_test, y_pred_baseline))
r2_baseline = r2_score(y_test, y_pred_baseline)

print("=== BASELINE ===")
print("MAE:", mae_baseline)
print("RMSE:", rmse_baseline)
print("R²:", r2_baseline)

print("\n=== RANDOM FOREST ===")
print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)
```

```python
import pandas as pd

# Obtener nombres de las variables después del OneHotEncoding
feature_names = (
    modelo.named_steps["preprocessor"]
    .get_feature_names_out()
)

# Importancias calculadas por Random Forest
importances = (
    modelo.named_steps["regressor"]
    .feature_importances_
)

feature_importance = pd.DataFrame({
    "variable": feature_names,
    "importancia": importances
}).sort_values("importancia", ascending=False)

feature_importance.head(15)
```

```python
top_features = feature_importance.head(15).sort_values("importancia")

plt.figure(figsize=(10, 7))

plt.barh(
    top_features["variable"],
    top_features["importancia"]
)

plt.xlabel("Importancia")
plt.ylabel("Variable")
plt.title("Variables más importantes para predecir producción de petróleo")

plt.tight_layout()
plt.show()
```

```python
# Separación temporal
train_temporal = df_ml[df_ml["mes"] <= 9]
test_temporal = df_ml[df_ml["mes"] >= 10]

X_train_temp = train_temporal.drop(columns=["prod_pet"])
y_train_temp = train_temporal["prod_pet"]

X_test_temp = test_temporal.drop(columns=["prod_pet"])
y_test_temp = test_temporal["prod_pet"]

print("Train:", X_train_temp.shape)
print("Test:", X_test_temp.shape)

print("\nMeses entrenamiento:")
print(sorted(X_train_temp["mes"].unique()))

print("\nMeses test:")
print(sorted(X_test_temp["mes"].unique()))
```

```python
modelo_temporal = Pipeline([
    ("preprocessor", preprocessor),
    ("regressor", RandomForestRegressor(
        n_estimators=100,
        max_depth=15,
        random_state=42,
        n_jobs=-1
    ))
])

print("Entrenando modelo temporal...")

modelo_temporal.fit(X_train_temp, y_train_temp)

print("Modelo temporal entrenado.")
```

```python
# Predicciones sobre meses futuros
y_pred_temp = modelo_temporal.predict(X_test_temp)

# Métricas
mae_temp = mean_absolute_error(y_test_temp, y_pred_temp)
rmse_temp = np.sqrt(mean_squared_error(y_test_temp, y_pred_temp))
r2_temp = r2_score(y_test_temp, y_pred_temp)

print("=== VALIDACIÓN TEMPORAL ===")
print("MAE:", mae_temp)
print("RMSE:", rmse_temp)
print("R²:", r2_temp)

print("\n=== SPLIT ALEATORIO ===")
print("MAE:", mae)
print("RMSE:", rmse)
print("R²:", r2)
```
